```python
import os

def list_mazes(directory):
    txt_files = []
    for filename in os.listdir(directory):
        if filename.endswith(".txt"):
            txt_files.append(filename)
    txt_files.sort()
    return txt_files

mazes = list_mazes(".")

for maze_file in mazes:
    with open(maze_file, "r") as f:
        maze_str = f.read()
    maze = mh.parse_maze(maze_str)
    
    print(maze_file)
    
    mh.show_maze(maze)
```

**1. Năm thành phần của bài toán tìm kiếm:**
*   **Trạng thái ban đầu (Initial state):** Vị trí bắt đầu của mình trong mê cung, chính là tọa độ chữ `S`.
*   **Tập hành động (Actions):** Tại mỗi ô, mình có thể di chuyển tối đa 4 hướng: Lên, Xuống, Trái, Phải (miễn là hướng đó không bị vướng tường).
*   **Mô hình chuyển trạng thái (Transition model):** Vị trí mới sau khi thực hiện một bước đi hợp lệ. Ví dụ, đang đứng ở (x, y) mà bước sang phải thì trạng thái mới sẽ là (x+1, y).
*   **Trạng thái đích (Goal test):** Điều kiện để dừng thuật toán, tức là kiểm tra xem tọa độ hiện tại đã trùng với chữ `G` (đích) hay chưa.
*   **Chi phí đường đi (Path cost):** Cứ mỗi bước đi sang ô kề bên sẽ tính là 1 đơn vị chi phí. Tổng chi phí đường đi chính là tổng số bước mình đã đi.

**2. Tính toán và ước lượng các đại lượng $n, d, m, b$:**
*(Lấy ví dụ minh họa trên bản đồ mẫu `empty_maze_2.txt`)*
*   **$n$ (Kích thước không gian trạng thái):** Là tổng số ô trống hợp lệ mà tác tử có thể đứng. Với mê cung kích thước viền 12x12, phần lõi di chuyển thực tế là 10x10, nên không gian trạng thái $n \approx 100$ trạng thái.
*   **$b$ (Hệ số phân nhánh tối đa):** Khi đứng ở vùng trống không có tường vây xung quanh, tác tử có thể rẽ sang tối đa 4 hướng, vậy hệ số phân nhánh lớn nhất $b = 4$.
*   **$d$ (Độ sâu lời giải tối ưu):** Số bước ít nhất để đi từ `S` đến `G`. Do mê cung rỗng, đường đi ngắn nhất chính là khoảng cách đường chim bay (khoảng cách Manhattan), ước lượng khoảng $d \approx 16$ bước.
*   **$m$ (Độ sâu tối đa):** Chiều dài đường đi dài nhất có thể mà không bị lặp lại ô cũ. Trong trường hợp xấu nhất đi vòng vèo qua toàn bộ các ô trống, chiều dài này sẽ tiệm cận tổng số ô. Vậy $m \le n \approx 100$.
